```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv('Iris.csv')
df.head()
```

```python
# Labels (for coloring)
labels = df['Species'].astype('category').cat.codes

# Features only (drop Id and Species)
X_features = df.drop(columns=['Id', 'Species'])
```

```python
df.shape
```

```python
pca = PCA(n_components=4)
X = pca.fit_transform(X_features)
```

```python
plt.figure(figsize=(6,6))

plt.scatter(X[:,0], X[:,1], c=labels, cmap='viridis')
plt.scatter(X[10,0], X[10,1], s=200, c='r', alpha=0.5)

plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA on Iris Dataset')
plt.colorbar(label='Class')
plt.show()
```
